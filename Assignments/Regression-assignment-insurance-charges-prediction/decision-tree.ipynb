```python
import pandas as pd
dataset=pd.read_csv(open("insurance_pre.csv",'rb'))
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent=dataset[['charges']]
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
x_train
x_test
```

```python
from sklearn.tree import DecisionTreeRegressor
#regressor = DecisionTreeRegressor(criterion='poisson',max_features=None,splitter='best')
#regressor = DecisionTreeRegressor(criterion='poisson',max_features=None,splitter='random')
#regressor = DecisionTreeRegressor(criterion='poisson',max_features='sqrt',splitter='best')
#regressor = DecisionTreeRegressor(criterion='poisson',max_features='sqrt',splitter='random')
#regressor = DecisionTreeRegressor(criterion='poisson',max_features='log2',splitter='best')
regressor = DecisionTreeRegressor(criterion='poisson',max_features='log2',splitter='random')


regressor=regressor.fit(x_train,y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
tree.plot_tree(regressor)
plt.show()
```

```python
y_pred = regressor.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
r_score
```

```python
import pickle
fileName ='finalize-decision-tree-model.sav'
load_model=pickle.dump(regressor,open(fileName,'wb'))
```
